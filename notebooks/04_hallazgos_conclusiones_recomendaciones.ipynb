# 04 · Hallazgos, conclusiones y recomendaciones

## 4.1 Propósito del análisis

Este documento reúne los principales resultados del dashboard **Desempeño Comercial y Operativo del Marketplace** y los interpreta en función de los objetivos definidos en el notebook 01. El análisis se dirige a la **gerencia de marketplace y operaciones** y busca orientar decisiones sobre descuentos, prioridades comerciales y desempeño de las entregas.

Los resultados corresponden a las **62.000 órdenes** de la base procesada, para el periodo comprendido entre el **1 de enero de 2024 y el 30 de septiembre de 2026**. Las cifras se presentan para el conjunto completo de órdenes, salvo cuando se indica una comparación temporal específica.

Los indicadores generales del dashboard muestran el siguiente panorama:

| Indicador | Resultado | Lectura para la gerencia |

## 4.3 Hallazgos sobre Categorías, fuentes de tráfico y medios de pago

**Objetivo:** identificar qué categorías concentran el ingreso y comparar cuántas órdenes llegan por cada fuente de tráfico y medio de pago.

### Categorías que concentran las ventas netas

Tecnología es la categoría de mayor peso económico dentro del dashboard: reúne **52,23 % de las ventas netas** con **21,18 % de las órdenes**. Su ticket promedio es de aproximadamente **610.971 COP**, el más alto entre las categorías analizadas. La participación en ventas es, por tanto, mucho mayor que su participación en el número de órdenes.

Hogar ocupa el segundo lugar con **15,24 % de las ventas netas**, mientras que Moda genera **19,84 % de las órdenes**, pero concentra **11,08 % de las ventas netas**. Esto evidencia que el volumen de órdenes y el valor generado no se distribuyen de la misma forma entre las categorías.

### Fuentes de tráfico

### Comportamiento temporal

La evolución mensual del dashboard muestra variaciones entre meses. En el periodo completo, las ventas netas mensuales van de **417,80 millones COP en febrero de 2025** a **500,70 millones COP en mayo de 2026**.

Al comparar enero-septiembre de cada año, para utilizar periodos equivalentes, las ventas netas fueron de **4.193,49 millones COP en 2024**, **4.162,25 millones en 2025** y **4.189,48 millones en 2026**. El resultado de enero-septiembre de 2026 se encuentra **0,10 % por debajo de 2024** y **0,65 % por encima de 2025**. En este periodo comparable, el valor neto se mantiene relativamente estable; los datos no muestran un crecimiento sostenido entre los tres años.

### Conclusión

Tecnología es la principal categoría por valor neto, mientras que Orgánico y Tarjeta son, respectivamente, la fuente de tráfico y el medio de pago con mayor número de órdenes. Estas medidas describen la composición de la operación, pero el número de órdenes por fuente o medio de pago no permite por sí solo concluir cuál canal es más rentable o cuál merece mayor inversión.


## 4.4 Hallazgos sobre Descuentos y devoluciones

**Objetivo:** evaluar si los descuentos generan más valor o si reducen el valor de cada compra y se relacionan con mayores devoluciones.

El valor de los descuentos aplicados suma **1.629,42 millones COP**, equivalente al **9,59 % del valor bruto de las órdenes**. Esta magnitud permite dimensionar el valor concedido mediante descuentos.

Al comparar los rangos extremos, el ticket bruto promedio pasa de **271.129 COP** en el rango de 0–5 % de descuento a **280.133 COP** en el rango de 20 % o más, un aumento aproximado de **3,32 %**. En contraste, el ticket neto promedio pasa de **266.852 COP** a **214.045 COP**, una disminución aproximada de **19,79 %**.

| Rango de descuento | Órdenes | Participación de las órdenes | Participación de las ventas netas | Ticket neto promedio (COP) | Devoluciones |
|---|---:|---:|---:|---:|---:|
| 0–5 % | 19.117 | 30,83 % | 33,22 % | 266.852 | 6,49 % |
| 5–10 % | 14.927 | 24,08 % | 24,78 % | 254.976 | 6,50 % |
| 10–15 % | 13.672 | 22,05 % | 21,38 % | 240.210 | 6,63 % |
| 15–20 % | 8.998 | 14,51 % | 13,25 % | 226.203 | 7,12 % |
| 20 % o más | 5.286 | 8,53 % | 7,37 % | 214.045 | 7,02 % |

El rango de 20 % o más reúne **8,53 % de las órdenes** y **7,37 % de las ventas netas**. En cambio, el rango de 0–5 % concentra 30,83 % de las órdenes y 33,22 % de las ventas netas. El ticket neto promedio disminuye a medida que se recorren los rangos de descuento, aunque el ticket bruto promedio no presenta una caída equivalente.

Las devoluciones no aumentan de forma continua en todos los rangos. La tasa más alta corresponde al rango de **15–20 % (7,12 %)**; en el de 20 % o más es de 7,02 %. Entre los rangos de 0–5 % y 20 % o más, la tasa pasa de 6,49 % a 7,02 %, una diferencia de aproximadamente **0,53 puntos porcentuales**.

Por categoría, Deportes registra el mayor descuento promedio (**9,62 %**), seguido de Moda (**9,58 %**). Las diferencias entre categorías son reducidas, por lo que conviene leerlas junto con la participación en ventas, el ticket promedio y las devoluciones.

### Conclusión

En los datos observados, los descuentos de mayor profundidad están asociados con un ticket neto promedio menor. La diferencia en devoluciones entre los rangos extremos es más pequeña que la caída del ticket y el máximo de devoluciones aparece en el rango de 15–20 %, no en el de 20 % o más. Por tanto, la evidencia respalda revisar la política de descuentos por rango y categoría, pero no permite afirmar que el descuento por sí solo cause más órdenes o devoluciones.


## 4.5 Hallazgos sobre Retrasos, calificación y devoluciones

**Objetivo:** medir cuánto disminuye la calificación del cliente a medida que aumentan los días de retraso e identificar dónde conviene actuar.

Del total de órdenes, **88,07 % se entregó sin retraso** y **11,93 % registró uno o más días de retraso**. La calificación promedio disminuye conforme aumenta el retraso: pasa de **4,31 sobre 5** en las órdenes sin retraso a **3,89 sobre 5** en las órdenes con cuatro días de retraso. La diferencia entre estos extremos es de aproximadamente **0,41 puntos**.

| Días de retraso | Calificación promedio (escala de 1 a 5) | Órdenes devueltas en 30 días |
|---:|---:|---:|
| 0 | 4,31 | 6,61 % |
| 1 | 4,19 | 6,83 % |
| 2 | 4,07 | 7,24 % |
| 3 | 3,98 | 6,97 % |
| 4 | 3,89 | 7,03 % |

La devolución alcanza su valor más alto en las órdenes con **dos días de retraso (7,24 %)**. Después disminuye y vuelve a subir ligeramente para cuatro días (7,03 %). Por ello, la relación entre retraso y devolución no sigue una progresión continua; el patrón más claro en el dashboard es el descenso de la calificación promedio conforme aumenta el número de días de retraso.

### Diferencias por operador y región

En la entrega a tiempo por operador, **Aliado logístico** presenta el mayor porcentaje (**88,24 %**) y **Propia** el menor (**87,88 %**). La brecha entre ambos es de aproximadamente **0,35 puntos porcentuales**.

Por región, **Orinoquía** registra el mayor porcentaje de entrega a tiempo (**88,57 %**) y **Suroccidente** el menor (**87,58 %**), con una diferencia cercana a **0,99 puntos porcentuales**. Estas diferencias permiten identificar segmentos para revisar, aunque su magnitud es moderada y no justifica por sí sola concluir que un operador o una región explica el desempeño general.

### Conclusión

La puntualidad es una prioridad operativa porque los retrasos afectan a casi 12 % de las órdenes y se observa una calificación promedio menor en los grupos con más días de retraso. La primera respuesta gerencial debería centrarse en reducir los retrasos y revisar el desempeño de operadores y regiones con tasas relativamente inferiores, sin asumir causalidad ni perder de vista la variación de las devoluciones entre grupos.


## 4.6 Conclusiones generales

1. **El valor neto está concentrado en Tecnología.** La categoría aporta 52,23 % de las ventas netas con 21,18 % de las órdenes. Su ticket promedio más alto explica parte importante de la diferencia entre participación en ventas y participación en volumen.

2. **Los descuentos más altos reducen el ticket neto promedio observado.** Entre los rangos de 0–5 % y 20 % o más, el ticket neto promedio disminuye 19,79 %, mientras que el ticket bruto promedio aumenta 3,32 %. Las devoluciones muestran diferencias más pequeñas y no crecen de manera continua con cada rango.

3. **Los retrasos se asocian con una menor calificación del producto recibido.** La calificación promedio pasa de 4,31 sin retraso a 3,89 con cuatro días de retraso. El desempeño puntual presenta diferencias entre operadores y regiones, pero la brecha entre los extremos es relativamente reducida.

4. **El comportamiento temporal comparable es estable.** Las ventas netas de enero a septiembre de 2026 quedan 0,10 % por debajo de las de 2024. Conviene seguir comparando periodos equivalentes para evitar interpretar como caída la falta de meses en un año incompleto.

En respuesta a la pregunta general del proyecto, la evidencia del dashboard sugiere priorizar la revisión de la política de descuentos por su valor monetario observable; abordar después la puntualidad de las entregas por su relación con la calificación; y mantener el seguimiento comercial de Tecnología, que concentra la mayor parte de las ventas netas.


## 4.7 Recomendaciones para la gerencia de marketplace y operaciones

### 1. Revisar los rangos de descuento de mayor profundidad

Evaluar con detalle las órdenes con descuentos de 15 % o más, especialmente las del rango de 20 % o más, porque presentan un ticket neto promedio considerablemente inferior al del rango de 0–5 %. La revisión debe hacerse por categoría y acompañarse del seguimiento de la participación en órdenes, ventas netas y devoluciones. Deportes y Moda son puntos de partida razonables para observar el descuento promedio, sin asumir que una diferencia pequeña implique por sí misma un problema.

### 2. Establecer la reducción de retrasos como prioridad operativa

Utilizar el 11,93 % de órdenes con retraso como indicador de seguimiento y revisar las diferencias por operador y región. Propia y Suroccidente muestran los menores porcentajes de entrega a tiempo en sus respectivos grupos; conviene identificar en la operación qué tipos de órdenes están asociados con esos resultados y verificar si las diferencias persisten en los periodos revisados. La prioridad es reducir los días de retraso y proteger la experiencia del cliente.

### 3. Proteger el desempeño comercial de Tecnología

Mantener seguimiento específico a Tecnología por su contribución de 52,23 % de las ventas netas y su ticket promedio superior al resto de categorías. Las decisiones comerciales deben considerar simultáneamente el valor aportado y el número de órdenes, evitando depender solo de la participación en volumen.

### 4. Usar las fuentes de tráfico y los medios de pago como línea base de seguimiento

Monitorear Orgánico y Pago por ser las fuentes con mayor número de órdenes, y Tarjeta y PSE/Transferencia por concentrar la mayor parte de los pagos. Antes de redistribuir inversión entre canales, complementar este seguimiento con información sobre costos de adquisición y conversión, variables que no están incluidas en el dataset utilizado.

### 5. Mantener comparaciones temporales equivalentes

Cuando se comparen años, utilizar la opción de enero-septiembre para mantener el mismo periodo. Las diferencias observadas son pequeñas y no justifican presentar 2026 como un año completo ni afirmar una tendencia de crecimiento sostenido.

### Orden recomendado de actuación

| Prioridad | Frente | Acción principal | Resultado a monitorear en el dashboard |
|---:|---|---|---|
| 1 | Descuentos | Revisar la profundidad de los descuentos por rango y categoría. | Ticket bruto y neto promedio, participación en ventas y tasa de devoluciones. |
| 2 | Entregas | Revisar las causas operativas de los retrasos, por operador y región. | Porcentaje de entrega a tiempo y calificación promedio según días de retraso. |
| 3 | Desempeño comercial | Mantener seguimiento a Tecnología y a la composición de las órdenes por canal y pago. | Participación por categoría, órdenes por fuente de tráfico y medio de pago. |
| 4 | Seguimiento gerencial | Comparar periodos equivalentes y segmentos con los filtros del dashboard. | Ventas netas mensuales, KPI y diferencias entre segmentos. |
