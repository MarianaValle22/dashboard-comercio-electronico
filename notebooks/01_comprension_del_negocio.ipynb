# 01 · Comprensión del negocio


## 1.1 Contexto y problema de negocio

El comercio electrónico, es decir, la compra y venta de productos por internet, ya hace parte de la vida diaria. Hoy basta con elegir un producto en una página web o una aplicación, pagarlo en línea o al recibirlo y esperar a que llegue a casa. Muchas veces todo este proceso ocurre en un *marketplace*, una plataforma que reúne en un mismo sitio los productos de muchos vendedores. Así se hizo una de cada cinco compras minoristas en el mundo en 2025 [1], y en Colombia las ventas en línea llegaron a $145,4 billones de pesos, cuatro veces más que en 2019 [2].

Sin embargo, vender por internet no termina cuando el cliente paga, y en cada etapa de la compra la plataforma puede perder dinero. Para atraer al cliente suele ofrecer descuentos, que aumentan las ventas pero reducen lo que se gana en cada una. Después, el pedido tiene que llegar a tiempo, ya que una entrega lenta es una de las razones más comunes por las que los clientes desisten de comprar [3]. Por último, si el producto no cumple lo que el cliente esperaba, este lo devuelve, algo que ocurre con casi uno de cada cinco productos comprados en línea [4]. Por eso, una plataforma puede vender mucho y, aun así, perder dinero y clientes.

Este proyecto analiza una plataforma nacional de comercio electrónico que enfrenta ese reto. La empresa no sabe con claridad en qué parte del proceso pierde valor: si en la venta (descuentos, categorías, canales por los que llegan los clientes, medios de pago) o en la entrega (retrasos, devoluciones, calificaciones bajas). La pregunta que guía el proyecto es: **¿dónde se pierden ingresos y satisfacción del cliente, y dónde conviene actuar primero?**

## 1.2 Audiencia y decisiones

El dashboard está dirigido a la **gerencia de marketplace y operaciones**, las personas que deciden sobre la política de descuentos, la mezcla de categorías, los canales para atraer clientes y el cumplimiento de las entregas.

Con el dashboard, esta audiencia podrá decidir:

- Dónde concentrar o moderar los descuentos.
- Qué categorías priorizar en la inversión comercial, según cuánto aporta cada una al ingreso.
- Cómo reducir los días de retraso en la entrega para mejorar la calificación del cliente, y si conviene enfocar ese esfuerzo en algún operador o región en particular.

## 1.3 Objetivos

### Objetivo general

Construir un dashboard interactivo que muestre a la gerencia de marketplace y operaciones dónde se gana y dónde se pierde valor a lo largo de la compra, desde la venta hasta la entrega, para orientar decisiones sobre descuentos, inversión comercial y operación logística.

### Objetivos específicos

> **Nota.** Los objetivos 1 y 3 se ajustaron tras la fase de comprensión de los datos (notebook 02).

1. Identificar qué categorías concentran el ingreso y comparar cuántas órdenes llegan por cada fuente de tráfico y medio de pago.
2. Evaluar si los descuentos generan más valor o si solo reducen el valor de cada compra y aumentan las devoluciones.
3. Medir cuánto disminuye la calificación del cliente a medida que aumentan los días de retraso en la entrega.
4. Presentar los hallazgos en un dashboard claro, con filtros que permitan a la gerencia explorar la información por su cuenta.

## 1.4 Recursos

El proyecto utiliza como única fuente el conjunto de datos asignado al grupo 10 (`data/raw/10_comercio_electronico.csv`), un registro sintético y anonimizado de la operación de una plataforma nacional de comercio electrónico. Su contenido y su calidad se revisan en la fase de comprensión de los datos.

## 1.5 Indicadores clave (KPI)

Para responder los objetivos específicos, el dashboard se apoyará en indicadores que resuman el desempeño de la plataforma a lo largo de la compra, desde la venta hasta la entrega. Su cálculo exacto se definirá una vez se conozca el contenido del conjunto de datos.

| Indicador | Qué mide | Para qué sirve |
|---|---|---|
| Ventas netas | Ingreso total de la plataforma después de descuentos. | Identificar qué categorías concentran el ingreso. |
| Ticket promedio | Valor promedio de cada compra. | Identificar qué categorías concentran el ingreso y evaluar el efecto de los descuentos. |
| Descuento promedio | Qué tan profundo es, en promedio, el descuento que se otorga. | Evaluar el efecto de los descuentos. |
| % entrega a tiempo | Proporción de pedidos entregados dentro del plazo prometido al cliente. | Medir el efecto de los retrasos. |
| % devoluciones | Proporción de pedidos que el cliente devuelve. | Evaluar el efecto de los descuentos. |
| Calificación promedio | Satisfacción del cliente con los productos que recibe. | Medir el efecto de los retrasos. |

## 1.6 Criterios de éxito

El proyecto será exitoso si señala, con cifras que lo respalden, en qué segmentos se pierde valor o satisfacción del cliente, y si la gerencia puede convertir esos hallazgos en al menos una decisión concreta sobre descuentos, inversión comercial u operación de entrega. Para lograrlo, el dashboard debe cumplir los objetivos específicos de la sección 1.3: mostrar qué sostiene el ingreso, qué efecto tienen los descuentos y cuánto afectan los retrasos en la entrega. Además, debe permitir explorar los datos con filtros, usar el color con un significado fácil de reconocer y leerse en orden, de lo general a lo particular.
